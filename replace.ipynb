```python
import json

# Load mappings
with open("mappings/kanz_mappings_english.json", "r", encoding="utf-8") as f:
    mappings = json.load(f)

# Reverse the mapping
reverse_mappings = {v: k for k, v in mappings.items()}

with open("dataset/train.txt", "r", encoding="utf-8") as infile, \
     open("dataset/train_updated.txt", "w", encoding="utf-8") as outfile:

    for line in infile:
        for old, new in reverse_mappings.items():
            line = line.replace(old, new)
        outfile.write(line)

print("Replacement complete.")
```

```python
import json

# Load mappings
with open("mappings/kanz_mappings_english.json", "r", encoding="utf-8") as f:
    mappings = json.load(f)

# Reverse the mapping
reverse_mappings = {v: k for k, v in mappings.items()}

with open("dataset/val.txt", "r", encoding="utf-8") as infile, \
     open("dataset/val_updated.txt", "w", encoding="utf-8") as outfile:

    for line in infile:
        for old, new in reverse_mappings.items():
            line = line.replace(old, new)
        outfile.write(line)

print("Replacement complete.")
```

```python
from collections import OrderedDict

chars = OrderedDict()

with open("dataset/train_updated.txt", encoding="utf8") as f:
    for line in f:
        _, text = line.strip().split("\t")
        for c in text:
            chars[c] = None

with open("dataset/custom_dict.txt", "w", encoding="utf8") as f:
    for c in chars:
        f.write(c + "\n")

print("Total characters:", len(chars))
```
