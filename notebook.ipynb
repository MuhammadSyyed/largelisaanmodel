```python
from pathlib import Path
from PIL import Image, ImageDraw, ImageFont
import arabic_reshaper
from bidi.algorithm import get_display

FONT_PATH = Path(
    r"C:\Users\ammar\OneDrive\Desktop\largelisaanmodel\Kanz-al-Marjaan\build\Kanz-al-Marjaan-Regular.ttf"
)

OUTPUT_DIR = Path("synthetic_dataset/images")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

sentences = [
    "الله تعالى ني نعمت",
    "اپنے مولا ني ذكر",
    "پہلے مؤمنين نے عرض كيدو",
    "لئے مؤمنين ني جماعت",
    "بسم الله الرحمن الرحيم",
]

font = ImageFont.truetype(str(FONT_PATH), size=58)


labels = []

for index, text in enumerate(sentences, start=1):
    reshaped = arabic_reshaper.reshape(text)
    display_text = get_display(reshaped)

    canvas = Image.new("L", (1600, 150), 255)
    draw = ImageDraw.Draw(canvas)

    bbox = draw.textbbox((0, 0), display_text, font=font)
    text_width = bbox[2] - bbox[0]

    x = 1550 - text_width
    y = 35

    draw.text((x, y), display_text, font=font, fill=0)

    image_path = OUTPUT_DIR / f"line_{index:05d}.png"
    canvas.save(image_path)

    labels.append(f"{image_path.as_posix()}\t{text}")

Path("synthetic_dataset/train.txt").write_text(
    "\n".join(labels),
    encoding="utf-8",
)
```
