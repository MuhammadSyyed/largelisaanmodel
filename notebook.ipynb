```python
from pathlib import Path
from fontTools.ttLib import TTFont
import json

FONT_PATH = Path(
    r"C:\Users\ammar\OneDrive\Desktop\largelisaanmodel\Kanz-al-Marjaan\build\Kanz-al-Marjaan-Regular.ttf"
)

font = TTFont(FONT_PATH)

# Glyph name -> Unicode character
glyph_to_unicode = {}
for codepoint, glyph in font.getBestCmap().items():
    glyph_to_unicode[glyph] = chr(codepoint)

gsub = font["GSUB"].table

mappings = {}

for lookup in gsub.LookupList.Lookup:

    # Only Ligature substitutions
    if lookup.LookupType != 4:
        continue

    for subtable in lookup.SubTable:

        for first, ligatures in subtable.ligatures.items():

            for lig in ligatures:

                # Build source sequence
                source = glyph_to_unicode.get(first, first)

                for g in lig.Component:
                    source += glyph_to_unicode.get(g, g)

                # Build target
                target = glyph_to_unicode.get(
                    lig.LigGlyph,
                    lig.LigGlyph
                )

                mappings[source] = target

# Save JSON
with open("kanz_mappings.json", "w", encoding="utf-8") as f:
    json.dump(
        mappings,
        f,
        ensure_ascii=False,
        indent=4
    )

print(f"Extracted {len(mappings)} mappings.")
print("Saved to kanz_mappings.json")
```

```python
def load_labels(file_path, split_name):
    records = []

    with open(file_path, "r", encoding="utf-8-sig") as f:
        for line_number, line in enumerate(f, start=1):

            line = line.rstrip("\r\n")

            if not line.strip():
                continue

            parts = line.split("\t", 1)

            if len(parts) != 2:
                print(
                    f"WARNING: malformed line "
                    f"{line_number} in {file_path.name}"
                )
                continue

            image_path, text = parts

            records.append({
                "split": split_name,
                "line_number": line_number,
                "image": image_path.strip(),
                "text_original": text
            })

    return pd.DataFrame(records)


train_df = load_labels(TRAIN_FILE, "train")
val_df   = load_labels(VAL_FILE, "val")
test_df  = load_labels(TEST_FILE, "test")

print("Train:", len(train_df))
print("Val  :", len(val_df))
print("Test :", len(test_df))
print("Total:", len(train_df) + len(val_df) + len(test_df))

display(train_df.head())
```

```python
import pandas as pd

ENGLISH_DIGITS = "0123456789"
ARABIC_DIGITS  = "٠١٢٣٤٥٦٧٨٩"
PERSIAN_DIGITS = "۰۱۲۳۴۵۶۷۸۹"

ALL_DIGITS = ENGLISH_DIGITS + ARABIC_DIGITS + PERSIAN_DIGITS


def find_numerals_in_txt(txt_path, split_name):
    results = []

    with open(txt_path, "r", encoding="utf-8") as f:

        for line_number, line in enumerate(f, start=1):

            line = line.rstrip("\n\r")

            if not line.strip():
                continue

            # Split ONLY on first tab
            parts = line.split("\t", 1)

            if len(parts) != 2:
                continue

            image_path, label = parts

            digits = "".join(
                c for c in label
                if c in ALL_DIGITS
            )

            if not digits:
                continue

            systems = []

            if any(c in ENGLISH_DIGITS for c in label):
                systems.append("English")

            if any(c in ARABIC_DIGITS for c in label):
                systems.append("Arabic")

            if any(c in PERSIAN_DIGITS for c in label):
                systems.append("Persian")

            results.append({
                "split": split_name,
                "line": line_number,
                "image": image_path,
                "label": label,
                "digits": digits,
                "numeral_system": ", ".join(systems),
            })

    return results
```

```python
def verify_images(df):
    bad_images = []

    for i, row in df.iterrows():
        path = row["full_image_path"]

        try:
            with Image.open(path) as img:
                img.verify()

        except Exception as e:
            bad_images.append({
                "image": row["image"],
                "error": str(e)
            })

    return pd.DataFrame(bad_images)


bad_train = verify_images(train_df)
bad_val   = verify_images(val_df)
bad_test  = verify_images(test_df)

print("Bad train images:", len(bad_train))
print("Bad val images  :", len(bad_val))
print("Bad test images :", len(bad_test))
```

```python
all_df = pd.concat(
    [train_df, val_df, test_df],
    ignore_index=True
)

character_counts = Counter()

for text in all_df["text_original"]:
    character_counts.update(text)


character_report = []

for char, count in character_counts.items():

    character_report.append({
        "character": char,
        "unicode": f"U+{ord(char):04X}",
        "name": unicodedata.name(char, "UNKNOWN"),
        "count": count
    })


char_df = pd.DataFrame(character_report)

char_df = char_df.sort_values(
    "count",
    ascending=False
).reset_index(drop=True)

print("Unique characters:", len(char_df))

display(char_df.head(100))
```

```python
def is_presentation_form(char):

    cp = ord(char)

    return (
        0xFB50 <= cp <= 0xFDFF or
        0xFE70 <= cp <= 0xFEFF
    )


presentation_chars = char_df[
    char_df["character"].apply(is_presentation_form)
]

print(
    "Unique Arabic Presentation Form characters:",
    len(presentation_chars)
)

display(presentation_chars)
```

```python
CUSTOM_KANZ_MAP = {
    "ظظ": "\uE001",
}

def apply_kanz_map(text):
    for source, target in CUSTOM_KANZ_MAP.items():
        text = text.replace(source, target)
    return text


def normalize_text(text):

    # 1. Convert English digits to Arabic-Indic digits
    text = text.translate(ENGLISH_TO_ARABIC_DIGITS)

    # 2. Apply Kanz-specific mappings before normalization
    text = apply_kanz_map(text)

    # 3. Normalize Arabic presentation forms
    text = unicodedata.normalize("NFKC", text)

    # 4. Apply Kanz mapping again in case NFKC creates the sequence
    text = apply_kanz_map(text)

    # 5. Clean whitespace
    text = " ".join(text.split())

    return text

changed_examples = []

for _, row in all_df.iterrows():

    original = row["text_original"]
    normalized = normalize_text(original)

    if original != normalized:

        changed_examples.append({
            "image": row["image"],
            "original": original,
            "normalized": normalized
        })


changed_df = pd.DataFrame(changed_examples)

print(
    "Labels affected by NFKC:",
    len(changed_df),
    "/",
    len(all_df)
)

display(changed_df.head(50))
```

```python
OCR_PROMPT = (
    "Transcribe the text in this image exactly. "
    "Preserve the original spelling, punctuation and characters. "
    "Do not translate, explain, correct or add any text. "
    "Return only the transcription."
)
```

```python
def create_qwen_record(row):

    return {
        "messages": [
            {
                "role": "user",
                "content": [
                    {
                        "type": "image",
                        "image": row["image"]
                    },
                    {
                        "type": "text",
                        "text": OCR_PROMPT
                    }
                ]
            },
            {
                "role": "assistant",
                "content": [
                    {
                        "type": "text",
                        "text": row["text_final"]
                    }
                ]
            }
        ]
    }
```

```python
def write_jsonl(df, output_file):

    with open(
        output_file,
        "w",
        encoding="utf-8"
    ) as f:

        for _, row in df.iterrows():

            record = create_qwen_record(row)

            f.write(
                json.dumps(
                    record,
                    ensure_ascii=False
                )
                + "\n"
            )


write_jsonl(
    train_df,
    OUTPUT_DIR / "train.jsonl"
)

write_jsonl(
    val_df,
    OUTPUT_DIR / "val.jsonl"
)

write_jsonl(
    test_df,
    OUTPUT_DIR / "test.jsonl"
)


print("Created:")
print(OUTPUT_DIR / "train.jsonl")
print(OUTPUT_DIR / "val.jsonl")
print(OUTPUT_DIR / "test.jsonl")
```

```python
# Final dataset sanity checks

for name, df in [
    ("train", train_df),
    ("val", val_df),
    ("test", test_df)
]:
    raw_kanz = df["text_final"].str.contains(
        "ظظ", regex=False, na=False
    ).sum()

    special_kanz = df["text_final"].str.contains(
        "\uE001", regex=False, na=False
    ).sum()

    english_digits = df["text_final"].apply(
        lambda x: any(c in "0123456789" for c in x)
    ).sum()

    arabic_digits = df["text_final"].apply(
        lambda x: any(c in "٠١٢٣٤٥٦٧٨٩" for c in x)
    ).sum()

    empty = (df["text_final"].str.len() == 0).sum()

    print(f"\n{name.upper()}")
    print("Rows:", len(df))
    print("Empty labels:", empty)
    print("Raw ظظ remaining:", raw_kanz)
    print("U+E001 samples:", special_kanz)
    print("English numeral rows:", english_digits)
    print("Arabic numeral rows:", arabic_digits)
```
